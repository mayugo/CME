```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Define symbols
s, t = sp.symbols('s t', real=True, positive=True)

# Define transfer function X(s)
X_s = 38 / ((6*s**2 + 2*s + 19)*s)

# Inverse Laplace transform
x_t = sp.inverse_laplace_transform(X_s, s, t)
```

```python
print("x(t) =")
sp.pprint(x_t)
```

```python
# Convert to numerical function
x_t_func = sp.lambdify(t, x_t, 'numpy')

# Time vector
t_vals = np.linspace(0, 20, 1000)

# Evaluate response
x_vals = x_t_func(t_vals)
```

```python
# Plot
plt.figure()
plt.plot(t_vals, x_vals)
plt.xlabel('t')
plt.ylabel('x(t)')
plt.title('Time Response x(t)')
plt.grid()
plt.show()
```
